```python
###Creating a Simple RDF Graph Visualizer

def show_rdf_graph(rdf_graph, title):

    visual_graph = nx.DiGraph()

    for subject, predicate, object_ in rdf_graph:

        subject_label = str(subject).split("/")[-1]
        predicate_label = str(predicate).split("/")[-1].split("#")[-1]
        object_label = str(object_).split("/")[-1]

        visual_graph.add_edge(
            subject_label,
            object_label,
            label=predicate_label
        )

    plt.figure(figsize=(14, 8))

    positions = nx.spring_layout(
        visual_graph,
        seed=42,
        k=1.5
    )

    nx.draw(
        visual_graph,
        positions,
        with_labels=True,
        node_size=3500,
        font_size=9,
        arrows=True
    )

    edge_labels = nx.get_edge_attributes(
        visual_graph,
        "label"
    )

    nx.draw_networkx_edge_labels(
        visual_graph,
        positions,
        edge_labels=edge_labels,
        font_size=8
    )

    plt.title(title)
    plt.axis("off")
    plt.show()
```

```python
###Creating a Reusable Graph Query Function

from rdflib import Literal

def get_projects_by_supervisor(graph, supervisor_name):

    query = """
    PREFIX utas: <https://example.org/utas-research/>
    PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>

    SELECT ?project ?title
    WHERE {
        ?project utas:supervisedBy ?supervisor .
        ?supervisor rdfs:label ?supervisorName .
        ?project utas:title ?title .

        FILTER(?supervisorName = ?wantedSupervisor)
    }
    """

    results = graph.query(
        query,
        initBindings={
            "wantedSupervisor": Literal(supervisor_name)
        }
    )

    return [
        {
            "project": str(row.project),
            "title": str(row.title)
        }
        for row in results
    ]
```

```python
###Creating a Tiny Question Planner

known_supervisors = [
    project["supervisor"]
    for project in projects
]

def mini_planner(question):

    question_lower = question.lower()

    for supervisor in known_supervisors:

        if supervisor.lower() in question_lower:

            if "project" in question_lower and "supervis" in question_lower:

                return {
                    "method": "graph",
                    "scope": "projects",
                    "graph_operation": "projects_by_supervisor",
                    "supervisor": supervisor,
                    "intent": "supervisor_project_lookup"
                }

    return {
        "method": "retrieval",
        "scope": "general",
        "search_query": question,
        "graph_operation": None,
        "supervisor": None,
        "intent": "general_information"
    }
```

```python
###Routing the Planner Decision to the Correct Graph Function

def mini_router(plan, graph):

    if plan["method"] == "graph":

        if plan["graph_operation"] == "projects_by_supervisor":

            result = get_projects_by_supervisor(
                graph,
                plan["supervisor"]
            )

            return {
                "route": "graph",
                "operation": "projects_by_supervisor",
                "result": result
            }

    return {
        "route": "retrieval",
        "operation": None,
        "result": None
    }
```

```python
###Creating a Simple Keyword Retriever

def keyword_retrieve(question, chunks):

    question_words = question.lower().split()

    scored_chunks = []

    for chunk in chunks:
        text = chunk["text"].lower()

        score = sum(
            1 for word in question_words
            if word in text
        )

        scored_chunks.append({
            "chunk_id": chunk["chunk_id"],
            "source": chunk["source"],
            "text": chunk["text"],
            "score": score
        })

    scored_chunks.sort(
        key=lambda item: item["score"],
        reverse=True
    )

    return scored_chunks
```

```python
###Combining BM25 and Semantic Rankings with RRF

question = "What English requirements are needed for research applicants?"

# BM25 ranking
bm25_scores = bm25.get_scores(
    question.lower().split()
)

bm25_ranked = sorted(
    range(len(general_chunks)),
    key=lambda i: bm25_scores[i],
    reverse=True
)

# Semantic ranking
question_embedding = model.encode([question])

semantic_scores = cosine_similarity(
    question_embedding,
    embeddings
)[0]

semantic_ranked = sorted(
    range(len(semantic_passages)),
    key=lambda i: semantic_scores[i],
    reverse=True
)

print("BM25 ranking:", bm25_ranked)
print("Semantic ranking:", semantic_ranked)
```

```python
###Combining Rankings with Reciprocal Rank Fusion

def rrf_score(rank, k=60):
    return 1 / (k + rank)

hybrid_scores = {}

for rank, index in enumerate(bm25_ranked, start=1):
    hybrid_scores[index] = hybrid_scores.get(index, 0) + rrf_score(rank)

for rank, index in enumerate(semantic_ranked, start=1):
    hybrid_scores[index] = hybrid_scores.get(index, 0) + rrf_score(rank)

hybrid_ranked = sorted(
    hybrid_scores,
    key=hybrid_scores.get,
    reverse=True
)

for index in hybrid_ranked:
    print(
        general_chunks[index]["chunk_id"],
        general_chunks[index]["source"],
        round(hybrid_scores[index], 5)
    )
```

```python
###Creating a Reusable Hybrid Retriever

def hybrid_retrieve(question, chunks, bm25_model, embedding_model, passage_embeddings):

    # BM25 ranking
    bm25_scores = bm25_model.get_scores(
        question.lower().split()
    )

    bm25_ranked = sorted(
        range(len(chunks)),
        key=lambda i: bm25_scores[i],
        reverse=True
    )

    # Semantic ranking
    question_embedding = embedding_model.encode([question])

    semantic_scores = cosine_similarity(
        question_embedding,
        passage_embeddings
    )[0]

    semantic_ranked = sorted(
        range(len(chunks)),
        key=lambda i: semantic_scores[i],
        reverse=True
    )

    # RRF
    hybrid_scores = {}

    for rank, index in enumerate(bm25_ranked, start=1):
        hybrid_scores[index] = (
            hybrid_scores.get(index, 0)
            + 1 / (60 + rank)
        )

    for rank, index in enumerate(semantic_ranked, start=1):
        hybrid_scores[index] = (
            hybrid_scores.get(index, 0)
            + 1 / (60 + rank)
        )

    best_index = max(
        hybrid_scores,
        key=hybrid_scores.get
    )

    return {
        "chunk_id": chunks[best_index]["chunk_id"],
        "source": chunks[best_index]["source"],
        "text": chunks[best_index]["text"],
        "hybrid_score": hybrid_scores[best_index]
    }
```

```python
###Updating the Hybrid Retriever to Return Top 2 Evidence

def hybrid_retrieve(
    question,
    chunks,
    bm25_model,
    embedding_model,
    passage_embeddings
):

    # BM25 ranking
    bm25_scores = bm25_model.get_scores(
        question.lower().split()
    )

    bm25_ranked = sorted(
        range(len(chunks)),
        key=lambda i: bm25_scores[i],
        reverse=True
    )

    # Semantic ranking
    question_embedding = embedding_model.encode([question])

    semantic_scores = cosine_similarity(
        question_embedding,
        passage_embeddings
    )[0]

    semantic_ranked = sorted(
        range(len(chunks)),
        key=lambda i: semantic_scores[i],
        reverse=True
    )

    # RRF
    hybrid_scores = {}

    for rank, index in enumerate(bm25_ranked, start=1):
        hybrid_scores[index] = (
            hybrid_scores.get(index, 0)
            + 1 / (60 + rank)
        )

    for rank, index in enumerate(semantic_ranked, start=1):
        hybrid_scores[index] = (
            hybrid_scores.get(index, 0)
            + 1 / (60 + rank)
        )

    # Final hybrid ranking
    hybrid_ranked = sorted(
        hybrid_scores,
        key=hybrid_scores.get,
        reverse=True
    )

    # Return top 2 chunks
    top_indices = hybrid_ranked[:2]

    return [
        {
            "chunk_id": chunks[index]["chunk_id"],
            "source": chunks[index]["source"],
            "text": chunks[index]["text"],
            "hybrid_score": hybrid_scores[index]
        }
        for index in top_indices
    ]
```

```python
###Converting Different Router Results into a Common Evidence Format

def prepare_evidence(router_output):

    if router_output["route"] == "graph":

        evidence_items = []

        for project in router_output["evidence"]:

            evidence_items.append({
                "source_type": "knowledge_graph",
                "source": project["project"],
                "text": project["title"]
            })

        return evidence_items

    if router_output["route"] == "retrieval":

        item = router_output["evidence"]

        return [
            {
                "source_type": "retrieved_text",
                "source": item["source"],
                "text": item["text"]
            }
        ]
```

```python
###Building the Prompt Sent to the LLM

def build_llm_prompt(question, evidence):

    evidence_text = ""

    for index, item in enumerate(evidence, start=1):
        evidence_text += (
            f"\nEvidence {index}\n"
            f"Source type: {item['source_type']}\n"
            f"Source: {item['source']}\n"
            f"Content: {item['text']}\n"
        )

    prompt = f"""
You are a research degree assistant.

Answer the user's question using only the evidence provided below.
If the evidence is insufficient, say that the available evidence is insufficient.
Do not invent information.

USER QUESTION:
{question}

EVIDENCE:
{evidence_text}

ANSWER:
"""

    return prompt
```

```python
###Creating a Qwen-Based Planner

import json
import requests

def qwen_planner(question):

    planner_prompt = f"""
You are a query planner for a research degree assistant.

Available methods:
- graph
- retrieval

Available graph operation:
- projects_by_supervisor

Known supervisors:
- Alice Smith
- Bob Lee

Rules:
- If the user asks which projects are supervised by a named supervisor,
  use method "graph".
- For general research degree information,
  use method "retrieval".

Return only valid JSON with these fields:

method
scope
graph_operation
supervisor
search_query
intent

User question:
{question}
"""

    response = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model": "qwen3:1.7b",
            "prompt": planner_prompt,
            "format": "json",
            "stream": False
        },
        timeout=120
    )

    response.raise_for_status()

    return json.loads(
        response.json()["response"]
    )
```

```python
###Adding a Safe Planner with Deterministic Fallback

def safe_planner(question, simulate_qwen_failure=False):

    try:
        if simulate_qwen_failure:
            raise RuntimeError("Simulated Qwen failure")

        plan = qwen_planner(question)

        plan["planner_method"] = "llm"

        return plan

    except Exception as error:

        print("Qwen planner unavailable:", error)
        print("Using deterministic fallback...\n")

        plan = mini_planner(question)

        plan["planner_method"] = "fallback"

        return plan
```

```python
###Validating the Planner Output Before Routing

ALLOWED_GRAPH_OPERATIONS = {
    "projects_by_supervisor"
}

def validate_plan(plan):

    if plan["method"] not in {"graph", "retrieval"}:
        raise ValueError("Unsupported method")

    if plan["method"] == "graph":

        if plan.get("graph_operation") not in ALLOWED_GRAPH_OPERATIONS:
            raise ValueError("Unsupported graph operation")

        if plan["graph_operation"] == "projects_by_supervisor":
            if not plan.get("supervisor"):
                raise ValueError("Supervisor name is required")

    if plan["method"] == "retrieval":
        if not plan.get("search_query"):
            raise ValueError("Search query is required")

    return plan
```

```python
###Integrating Planner Validation with Automatic Fallback

def safe_planner(question, simulate_qwen_failure=False):

    try:
        # Simulating Qwen failure when needed for testing
        if simulate_qwen_failure:
            raise RuntimeError("Simulated Qwen failure")

        # Getting the plan from Qwen
        plan = qwen_planner(question)

        # Validating the Qwen plan
        plan = validate_plan(plan)

        plan["planner_method"] = "llm"

        return plan

    except Exception as error:

        print("Qwen planner failed or produced an invalid plan:")
        print(error)

        print("\nUsing deterministic fallback...\n")

        # Creating the deterministic fallback plan
        plan = mini_planner(question)

        # Validating the fallback plan as well
        plan = validate_plan(plan)

        plan["planner_method"] = "fallback"

        return plan
```

```python
###Creating a Simple Evidence Guard

def guard_evidence(evidence):

    if not evidence:
        return False

    for item in evidence:
        if item.get("text", "").strip():
            return True

    return False
```

```python
###Adding Sources to the LLM Prompt

def build_llm_prompt(question, evidence):

    evidence_text = ""

    for index, item in enumerate(evidence, start=1):

        evidence_text += (
            f"\nEvidence {index}\n"
            f"Source: {item['source']}\n"
            f"Content: {item['text']}\n"
        )

    prompt = f"""
You are a research degree assistant.

Answer using only the evidence provided below.

Rules:
- Do not invent information.
- If the evidence is insufficient, say so.
- Mention the source used in the answer.

USER QUESTION:
{question}

EVIDENCE:
{evidence_text}

ANSWER:
"""

    return prompt
```

```python
###Adding a Relevance Threshold to Hybrid Retrieval

def hybrid_retrieve(
    question,
    chunks,
    bm25_model,
    embedding_model,
    passage_embeddings,
    relevance_threshold=0.35
):

    # Semantic similarity
    question_embedding = embedding_model.encode([question])

    semantic_scores = cosine_similarity(
        question_embedding,
        passage_embeddings
    )[0]

    best_semantic_score = float(max(semantic_scores))

    # Stop if the question is not relevant to the knowledge base
    if best_semantic_score < relevance_threshold:
        return []

    # BM25 ranking
    bm25_scores = bm25_model.get_scores(
        question.lower().split()
    )

    bm25_ranked = sorted(
        range(len(chunks)),
        key=lambda i: bm25_scores[i],
        reverse=True
    )

    # Semantic ranking
    semantic_ranked = sorted(
        range(len(chunks)),
        key=lambda i: semantic_scores[i],
        reverse=True
    )

    # RRF
    hybrid_scores = {}

    for rank, index in enumerate(bm25_ranked, start=1):
        hybrid_scores[index] = (
            hybrid_scores.get(index, 0)
            + 1 / (60 + rank)
        )

    for rank, index in enumerate(semantic_ranked, start=1):
        hybrid_scores[index] = (
            hybrid_scores.get(index, 0)
            + 1 / (60 + rank)
        )

    hybrid_ranked = sorted(
        hybrid_scores,
        key=hybrid_scores.get,
        reverse=True
    )

    top_indices = hybrid_ranked[:2]

    return [
        {
            "chunk_id": chunks[index]["chunk_id"],
            "source": chunks[index]["source"],
            "text": chunks[index]["text"],
            "hybrid_score": hybrid_scores[index],
            "semantic_score": float(semantic_scores[index])
        }
        for index in top_indices
    ]
```
